Every pipeline you ever build will fail. The question is never whether — it is whether you
find out from a run history, or from somebody in a meeting.

This notebook fails on purpose, so that you can see what a job does about it.

```python
dbutils.widgets.dropdown("should_fail", "yes", ["yes", "no"], "Fail this run?")

# The dropdown appears at the TOP of the notebook, not below this cell.
print("A dropdown has been created at the TOP of this notebook: 'Fail this run?'")
print("Leave it on yes for now.")
```

```python
import datetime

print("starting the step that talks to the payment gateway...")
print(f"time now (UTC): {datetime.datetime.now(datetime.timezone.utc):%Y-%m-%d %H:%M:%S}")

if dbutils.widgets.get("should_fail") == "yes":
    raise RuntimeError("Payment gateway did not respond. This failure is deliberate.")

print("gateway responded. Step complete.")
```

| The task output | the actual error message and the line it came from |
| The run history | whether this is the first time, or the fourth night in a row |
| Your inbox | whether anybody was told |
