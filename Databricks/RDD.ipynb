```python
%sql

SELECT
    SUM(CAST(amount_inr AS DECIMAL(14,2))) AS total_decimal,
    SUM(CAST(amount_inr AS DOUBLE))        AS total_double,
    SUM(CAST(amount_inr AS FLOAT))         AS total_float
FROM workspace.default.upi_transactions_2026;
```

```python
%sql

SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(
        CAST(CAST(amount_inr AS FLOAT) AS DOUBLE)
        <> CAST(amount_inr AS DOUBLE)
    ) AS rows_no_longer_correct
FROM workspace.default.upi_transactions_2026;
```
